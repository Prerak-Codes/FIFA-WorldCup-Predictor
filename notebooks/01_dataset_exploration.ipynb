```python
import pandas as pd
import numpy as np
```

```python
results = pd.read_csv("../data/raw/international_results/results.csv")

print("=" * 60)
print("RESULTS DATASET")
print("=" * 60)

display(results.head())

print("\nShape:")
print(results.shape)

print("\nColumns:")
print(results.columns.tolist())

print("\nData Types:")
results.info()

print("\nStatistical Summary:")
display(results.describe(include="all"))

print("\nMissing Values:")
display(results.isnull().sum())

print("\nDuplicate Rows:")
print(results.duplicated().sum())

print("\nUnique Home Teams:")
print(results["home_team"].nunique())

print("\nUnique Away Teams:")
print(results["away_team"].nunique())

print("\nTournament Types:")
display(results["tournament"].value_counts())

print("\nDate Range:")
print(results["date"].min())
print(results["date"].max())
```
