```python
import pandas as pd

team_form = pd.read_csv("../data/raw/match_features/teams_form.csv")

team_form.head()
```

```python
print("Shape:")
print(team_form.shape)

print("\nColumns:")
print(team_form.columns.tolist())

print("\nData Types:")
team_form.info()
```

```python
print("\nMissing Values:")
display(team_form.isnull().sum())

print("\nDuplicate Rows:")
print(team_form.duplicated().sum())
```

```python
team_form.describe(include="all")
```
