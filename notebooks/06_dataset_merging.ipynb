```python
results["date"] = pd.to_datetime(results["date"])
elo["date"] = pd.to_datetime(elo["date"])
rankings["rank_date"] = pd.to_datetime(rankings["rank_date"])
match_features["_date"] = pd.to_datetime(match_features["_date"])
```

```python
elo.rename(columns={
    "team": "elo_team"
}, inplace=True)
```

```python
rankings.rename(columns={
    "team": "ranking_team",
    "rank_date": "date"
}, inplace=True)
```

```python
match_features.rename(columns={
    "_home_team": "home_team",
    "_away_team": "away_team",
    "_date": "date",
    "_tournament": "tournament"
}, inplace=True)
```

```python
print(results.columns)
print(elo.columns)
print(rankings.columns)
print(match_features.columns)
```

```python
merged = results.merge(
    match_features,
    on=["date", "home_team", "away_team", "tournament"],
    how="left"
)

print(merged.shape)
merged.head()
```
